### 1. Dictionary vs Typed Dictionary

```python
from typing import TypedDict

# Dictionary: Doesn't check if the data in the correct type or structure
movie_dict = {
    "name": "Avengers: Endgame",
    "year": 2019
}
print(f"Dictionary: {movie_dict}")

# Typed Dictionary: Type safety and structure enforcement
class Movie(TypedDict):
    name: str
    year: int
movie = Movie(name="Avengers: Endgame", year=2019)
print(f"Typed Dictionary: {movie}")
```

### 2. Union

```python
from typing import Union

# Union allows a variable to hold multiple types like: int, float etc.
def square(x: Union[int, float]) -> float:
    return x * x

x = 5
x = 1.234
print(square(x))

x = "hello"
print(square(x)) # This will raise a type checker warning
```

### 3. Optional

```python
from typing import Optional

# Optional indicates that a variable can either be of a specified type or None
def nice_message(name: Optional[str]) -> None:
    if name is None:
        print("Hello, Guest!")
    else:
        print(f"Hello, {name}!")
nice_message("Alice")
nice_message(None)
```

### 4. Any

```python
from typing import Any

# Any allows a variable to hold any type without type checking
def print_value(x: Any):
    print(x)
print_value(1)
print_value("Hello")
print_value([1, 2, 3])
```

### 5. Lambda Function

```python
# lambda: shortcut to writing small anonymous functions
square = lambda y: y * y
print(square(10))

nums = [1, 2, 3, 4]
squares = list(map(lambda x: x * x, nums))
print(squares)
```
